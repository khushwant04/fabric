# Fabric paper figures

This notebook regenerates the paper's architecture and evidence figures from the checked-in JSON evidence register. It does not synthesize missing measurements.

```python
from pathlib import Path
import json

repo_root = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / 'paper' / 'data' / 'evidence.json').exists()
)
evidence_path = repo_root / 'paper' / 'data' / 'evidence.json'
evidence = json.loads(evidence_path.read_text())
evidence['provenance']
```

```python
import runpy
runpy.run_path(str(repo_root / 'paper' / 'generate_figures.py'), run_name='__main__')
```

Generated vector PDFs and review PNGs are written to `paper/figures/`.
